# KAPP: Kenya Agricultural Production Prediction

## 1. Project Introduction

Agriculture plays a significant role in Kenya's economy, food security, employment, and rural livelihoods. However, agricultural production is influenced by several factors, including rainfall patterns, temperature variations, cultivated land area, and agricultural inputs.

The Kenya Agricultural Production Prediction (KAPP) project aims to develop a data-driven predictive system that uses historical agricultural and climatic data to estimate crop production across Kenyan counties.

The initial phase of the project will focus on maize production because of its importance to food security in Kenya.

The project will apply statistical analysis and machine learning techniques to identify production patterns, understand relationships between agricultural and climatic variables, and develop predictive models.

The ultimate goal is to provide useful insights that can support agricultural planning, resource allocation, and food security decision-making.

## 2. Problem Statement

Agricultural production in Kenya is affected by changing climatic conditions, unpredictable rainfall patterns, temperature fluctuations, variations in cultivated land, and differences in agricultural practices across counties.

These factors make it difficult for farmers, policymakers, and agricultural institutions to accurately anticipate future crop production.

Inadequate production forecasting can contribute to challenges in food security planning, agricultural resource allocation, market preparation, and national food supply management.

Although historical agricultural and climatic data are available from various institutions, there is a need to investigate how these datasets can be integrated and used to develop reliable predictive models.

KAPP seeks to address this challenge by applying statistical methods and machine learning algorithms to historical agricultural and climatic data to predict maize production across Kenyan counties.

The project will investigate the relationship between agricultural production and relevant influencing factors to establish how data-driven forecasting can support better agricultural decision-making.

```python
raw = pd.read_excel("../data/raw/Maize-Production-2012-2020-Combined.xlsx", header=None)

display(raw.head(8))
```

```python
import pandas as pd

counties = raw.iloc[3:, 0].str.strip().reset_index(drop=True)
data = raw.iloc[3:].reset_index(drop=True)

blocks = []
for start in range(1, raw.shape[1], 3):          # 1, 4, 7, ... 22
    year = int(raw.iloc[1, start])               # the year sits above the first column of each triple
    block = data.iloc[:, start:start + 3].copy()
    block.columns = ["harvested_area_ha", "production_mt", "yield_mt_ha"]
    block.insert(0, "county", counties)
    block.insert(1, "year", year)
    blocks.append(block)

df = pd.concat(blocks, ignore_index=True)

num_cols = ["harvested_area_ha", "production_mt", "yield_mt_ha"]
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")

print(df.shape)
print(df["county"].unique())
print(df.isna().sum())
```

```python
# 1. Every county has exactly 8 years
print(df.groupby("county")["year"].nunique().value_counts())

# 2. Does yield match production / area?
calc = df["production_mt"] / df["harvested_area_ha"]
diff = (calc - df["yield_mt_ha"]).abs()
print("max yield mismatch:", diff.max())
print(df[diff > 0.01])

# 3. Zeros and odd values
print((df[["harvested_area_ha", "production_mt", "yield_mt_ha"]] <= 0).sum())
print(df.describe().T)

# 4. Save
df.to_csv("../data/processed/maize_county_2012_2019.csv", index=False)
```

```python
import matplotlib.pyplot as plt

# National trend
df.groupby("year")["production_mt"].sum().plot(marker="o", title="Total maize production (MT)")
plt.show()

# Top producers
top = df.groupby("county")["production_mt"].mean().nlargest(10)
top.sort_values().plot.barh(title="Top 10 counties, mean production (MT)")
plt.show()

# Distribution (it's likely right-skewed, so check the log)
import numpy as np
fig, ax = plt.subplots(1, 2, figsize=(10, 3))
df["production_mt"].hist(ax=ax[0], bins=30)
np.log1p(df["production_mt"]).hist(ax=ax[1], bins=30)
plt.show()
```

 ## Lagged Feature Validation

The lagged feature validation identified an additional consideration arising from the missing 2019 observations.

The dataset contains two lagged variables:

- "prod_lag1": Previous-year maize production.
- "area_lag1": Previous-year harvested area.

Since 2019 is unavailable, the 2020 observations cannot obtain valid one-year lag values from the dataset.

The affected 2020 lagged values were therefore set to missing (NaN) rather than incorrectly using 2018 observations.

Following this adjustment, each lagged feature contains 94 missing values:

- 47 from the initial observation for each county.
- 47 from the 2020 observations.

This adjustment preserves the correct temporal relationship between observations.
